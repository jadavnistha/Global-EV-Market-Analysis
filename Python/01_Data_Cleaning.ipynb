```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned dataset
df = pd.read_csv("../Data/Clean/global_ev_sales_clean.csv")

# Sum EV sales for each year
yearly_sales = df.iloc[:, 1:].sum()

# Plot
plt.figure(figsize=(12,6))

plt.plot(
    yearly_sales.index,
    yearly_sales.values,
    marker='o',
    linewidth=3
)

plt.title("Global EV Sales Trend (2010–2024)", fontsize=16, fontweight="bold")
plt.xlabel("Year")
plt.ylabel("Total EV Sales")
plt.grid(True)

plt.tight_layout()

# Save chart
plt.savefig("../Data/Clean/ev_sales_trend.png", dpi=300)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned dataset
makers_df = pd.read_csv("../Data/Clean/makers_sales_clean.csv")

# Top 10 makers by 2024 sales
top10 = makers_df.sort_values(by="2024", ascending=False).head(10)

# Plot
plt.figure(figsize=(12,6))

plt.barh(top10["Maker"], top10["2024"])

plt.title("Top 10 EV Manufacturers by Sales (2024)", fontsize=16, fontweight="bold")
plt.xlabel("EV Sales")
plt.ylabel("Manufacturer")

plt.gca().invert_yaxis()  # Highest at the top

plt.tight_layout()

# Save chart
plt.savefig("../Data/Clean/top10_ev_makers_2024.png", dpi=300)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned dataset
vehicle_df = pd.read_csv("../Data/Clean/vehicle_class_clean.csv")

# Plot
plt.figure(figsize=(8,8))

plt.pie(
    vehicle_df["Total Registration"],
    labels=vehicle_df["Vehicle Class"],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.45)   # Makes it a donut chart
)

plt.title("EV Registration by Vehicle Class", fontsize=16, fontweight="bold")

# Save chart
plt.savefig("../Data/Clean/vehicle_class_donut.png", dpi=300)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned dataset
df = pd.read_csv("../Data/Clean/global_ev_sales_clean.csv")

# Top 10 countries by 2024 EV sales
top10 = df.sort_values(by="2024", ascending=False).head(10)

# Plot
plt.figure(figsize=(12,6))

plt.bar(top10["region_country"], top10["2024"])

plt.title("Top 10 Countries by EV Sales (2024)", fontsize=16, fontweight="bold")
plt.xlabel("Country")
plt.ylabel("EV Sales")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

# Save chart
plt.savefig("../Data/Clean/top10_countries_2024.png", dpi=300)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned dataset
maker_place_df = pd.read_csv("../Data/Clean/maker_place_clean.csv")

# Count manufacturers by State
state_count = (
    maker_place_df.groupby("State")
    .size()
    .sort_values(ascending=False)
)

# Plot
plt.figure(figsize=(10,6))

plt.bar(state_count.index, state_count.values)

plt.title("Top States by EV Manufacturers", fontsize=16, fontweight="bold")
plt.xlabel("State")
plt.ylabel("Number of EV Manufacturers")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

# Save chart
plt.savefig("../Data/Clean/top_states_ev_manufacturers.png", dpi=300)

plt.show()
```
